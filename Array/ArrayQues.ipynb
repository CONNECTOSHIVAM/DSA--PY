```python
# Q1. Given an array, check if it contains any duplicates or not.
# arr = [1, 2, 4, 2, 5, 9]
# Output = True

def check_duplicate(arr):
    l1 = set(arr)
    if l1 is arr:
        return False
    else:
        return True
    
#Driver Code
arr = [1, 2, 4, 2, 5, 9]
result = check_duplicate(arr)
print(result)
```

```python
# Q2. Given an array and an integer k, rotate the array to the right by k steps.
# arr = [1, 2, 3, 4, 5, 6, 7] ,k = 3
# Output = [5, 6, 7, 1, 2, 3, 4]

# arr = [1, 2, 3, 4, 5, 6, 7] 
# k=3
# print(arr[-3:]+arr[:-3])

arr = [1, 2, 3, 4, 5, 6, 7] 
k = 3
a = k%len(arr)
for i in range(a):
    b = arr.pop()
    arr.insert(0,b)
    
print(arr)
```
