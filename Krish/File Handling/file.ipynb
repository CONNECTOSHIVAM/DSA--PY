```python
# Q1. Write a function that reads the contents of a file named `sample.txt` and prints each line.

def read_file(filename):
    with open(filename,'r') as file:
        for line in file:
            print(line.strip())
            
print(read_file('sample.txt'))
```

```python
# Q2. Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

def write_on_file(filename,lst):
    with open(filename,'w') as file:
        for i in lst:
            file.write(i + '\n')
            
print(write_on_file('output.txt',["Be","Happy","and","Be","positive"]))
```

```python
# Q3.Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

def copy_file(source, destination):
    
    with open(source,'r') as src:
        with open(destination, 'w') as dest:
            dest.write(src.read())
            
print(copy_file('source.txt','destination.txt'))
```

```python
# Q4. Write a function that appends a given string to the end of a file named `log.txt`.

def append_to_file(filename, text):
    with open(filename, 'a') as file:
        file.write(text + '\n')
        
print(append_to_file('log.txt',"Hi , I m Genius"))
print(append_to_file('log.txt',"Hi , I m ConnectoShivam"))
```

```python
# Q5. Write a function that reads the contents of a file named `document.txt` and returns the number of words in the
# file.

def read_file(filename):
    with open(filename,'r') as file:
        contents = file.read()
        word = contents.split()
        return len(word)
    
print(read_file('log.txt'))
```

```python
# Q6. Write a function that finds and replaces all occurrences of a given word in a file named `log.txt` with
# another word.

def replace_all(filename,old_word,new_word):
    with open(filename,'r') as file:
        content = file.read()
    new_content = content.replace(old_word,new_word)
    with open(filename,'w') as file:
        file.write(new_content)
        
print(replace_all('log.txt','Genius','Unstopable'))
        
        
```

```python
# Q7. Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.

def reverse_contents(filename):
    
    with open(filename,'r') as file:
        text = file.readlines()
    for i in reversed(text):
        print(i.strip())
        
print(reverse_contents('log.txt'))
            

        
```

```python
# Q8.Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, 
# and characters in the file.

def count_wlc(filename):
    
    with open(filename,'r') as file:
        content = file.readlines()
        lines = len(content)
        words = sum(len(line.split()) for line in content)
        character = sum(len(line) for line in content)
    
    return lines, words,character
        

print(count_wlc('source.txt'))
```

```python
# Q9. Write a function that merges the contents of multiple files into a single file named `merged.txt`.

def merge_file(file_list,output_file):
    with open(output_file,'w') as outfile:
        for file in file_list:
            with open(file,'r') as infile:
                outfile.write(infile.read()+ '\n')
                
print(merge_file(['log.txt','source.txt'],'merged.txt'))
```

```python
# Q10. Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

def split_file(large_file,lines_per_file):
    
    with open(large_file,'r') as file:
        lines = file.readlines()
    for line in range(0,len(lines),lines_per_file):
        with open(f'{large_file}_part{line//lines_per_file + 1}.txt','w') as part_file:
            part_file.writelines(lines[line: line + lines_per_file])
            
print(split_file('large.txt',10))
```

```python
# Q11.Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

import datetime

def write_msg(message,filename):
    
    timestamps = datetime.datetime.now().isoformat()
    with open(filename,'a') as file:
        file.write(f'{timestamps}: {message}\n')
        
print(write_msg("Hi , I m log file",'activity.log'))
        
        
```

```python
# Q12. Write a function that reads a binary file named `image.bin` and writes its contents to another binary file
# named `copy_image.bin`.

def read_write(fileone, filetwo):
    with open(fileone,'rb') as file:
        content = file.read()
        with open(filetwo,'wb') as dest:
            dest.write(content)
            
            
print(read_write('image.bin','copy_image.bin'))
        
```

```python
# Q13.Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

import csv

def read_csv(filename):
    with open(filename,'r') as file:
        documents = csv.DictReader(file)
        return list(documents)
    
print(read_csv('data.csv'))
            
        
```

```python
# Q15. Write a function that attempts to read a file named `protected.txt` and handles any permission errors 
# gracefully by printing an error message.

def read_file(filename):
    try:
        with open(filename,'r') as file:
            print(file.read())
    except PermissionError as e:
        print(f"Permission error: {e}")
        

print(read_file('protected.txt'))
```
