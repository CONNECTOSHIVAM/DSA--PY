# Questions based on dictionary

```python
# Q1.Create a dictionary with the first 10 positive integers as keys and their squares as values. Print the dictionary.

my_dict = {i:i**2 for i in range(1,11)}
print(my_dict)
```

```python
# Q2. Print the value of the key 5 and the keys of the dictionary created in Question 1.

print(my_dict[5])
l1 = []
for i in my_dict:
    l1.append(i)
    
print(f"All key in dictionary: {l1}")
```

```python
# Q3. Add a new key-value pair (11, 121) to the dictionary created in Question 1 and then remove the key-value pair
# with key 1. Print the modified dictionary.

my_dict = {i:i**2 for i in range(1,11)}
my_dict[11]=121
print(my_dict)
my_dict.pop(1)
print('modified dictionary: ',my_dict)

```

```python
# Q4.Iterate over the dictionary created in Assignment 1 and print each key-value pair.

for i , j in my_dict.items():
    print(f'Key: {i} | Value: {j}')
```

```python
# Q5.Create a new dictionary containing the cubes of the first 10 positive integers using a dictionary comprehension. 
# Print the new dictionary.

my_cube = {i:i**3 for i in range(1,11)}
print(my_cube)
```

```python
# Q6.Create two dictionaries: one with keys as the first 5 positive integers and values as their squares, and another
# with keys as the next 5 positive integers and values as their squares. Merge these dictionaries into a 
# single dictionary and print it.

dict_one = {i:i**2 for i in range(1,6)}
dict_two = {i:i**2 for i in range(6,11)}

dict_one.update(dict_two)
print(dict_one)
```

```python
# Q7.Create a nested dictionary representing a student with keys 'name', 'age', 'grades', where 'grades' is another 
# dictionary with keys 'math', 'science', and 'english'. Print the nested dictionary.

student = {'name': 'shivam', 'age':21, 'grades': {'english': 98,'science':99,'math': 100} }
print(student)
```

```python
# Q8.Create a dictionary where the keys are the first 5 positive integers and the values are lists containing the 
# first 5 multiples of the key. Print the dictionary.

my_dict = {i:[i*j for j in range(1,6)] for i in range(1,6)}
print(my_dict)
```

```python
# Q9.Create a dictionary where the keys are the first 5 positive integers and the values are tuples containing 
# the key and its square. Print the dictionary.

my_dict = {i:(i,i**2) for i in range(1,6)}
print(my_dict)
```

```python
#Q10.Create a dictionary with the first 5 positive integers as keys and their squares as values. 
# Convert the dictionary to a list of tuples and print it.

my_dict = {i:i**2 for i in range(1,6)}
l1 = list(my_dict.items())
print(l1)
```

```python
# Q11.Create a dictionary with the first 10 positive integers as keys and their squares as values. Create a 
# new dictionary containing only the key-value pairs where the key is even. Print the new dictionary.

my_dict = {i:i**2 for i in range(1,11)}
d1 = {k:v for k,v in my_dict.items() if k%2==0}
print(d1)
```

```python
# Q12.Create a dictionary with the first 5 positive integers as keys and their squares as values. Create a new 
# dictionary with keys and values swapped. Print the new dictionary.

my_dict = {i:i**2 for i in range(1,11)}
print(my_dict)
new_dict = { j:i for i,j in my_dict.items()}
print(new_dict)
```

```python
# Q14.Write a function that takes a string and returns a dictionary with the count of each character in the string.
# Print the dictionary.

def myFunc(hello):
    my_dict = {}
    for i in hello:
        my_dict[i]=my_dict.get(i,0)+1
    return my_dict
        
        
hello = "Hi , I am Ms Dhoni"
print(myFunc(hello))
```
