# Question on Based Sets

```python
# Q1.Create a set with the first 10 positive integers. Print the set.

my_set = set(range(1,11))
print(my_set)
```

```python
# Q2. Add the number 11 to the set created in Question 1. Then remove the number 1 from the set. 
# Print the modified set.
my_set.add(11)
print(my_set)
my_set.remove(1)
print(my_set)
```

```python
# Q3.Create two sets: one with the first 5 positive integers and another with the first 5 even integers. 
# Perform and print the results of union, intersection, difference, and 
# symmetric difference operations on these sets.

my_set1 = set(range(1,6))
my_set2 = set(range(2,11,2))

print(my_set1)
print(my_set2)
union_set = my_set1.union(my_set2) #Union : All data of merge / add of sets.
print(f'Union set {union_set}') 
intersection_set = my_set1.intersection(my_set2) #Intersection : common data in both sets.
print(f"Intersection set: {intersection_set}")
difference_set = my_set1.difference(my_set2) #diffrence : unique elements in set1 one that elemensts are not avalaible in set2.
print(f"Difference set: {difference_set}") 
symmetric_difference_set = my_set1.symmetric_difference(my_set2) # symmetric_difference : All unique elements of both set -- not match elemets in adjecents sets.
print(f"Symmetric diffrence: {symmetric_difference_set}")
```

```python
# Q4.Create a new set containing the squares of the first 10 positive integers using a set comprehension.
# Print the new set.

my_set = set(range(1,11))
print(my_set)
square_set = {i**2 for i in my_set}
print(square_set)
```

```python
# Q5. Create a new set containing only the even numbers from the set created in Question 1 using a set comprehension.
# Print the new set.

my_set = set(range(1,6))
even_set = {i for i in my_set if i%2==0}
print(even_set)
```
