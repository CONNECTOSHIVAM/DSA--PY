```python
# Q1. Write a function that takes two integers as input and returns their division. Use try, except, and finally 
# blocks to handle division by zero and print an appropriate message.

def myFunc(a,b):
    try:
        result = a/b
    except ZeroDivisionError as e:
        print(e)
        result = None
    except Exception as ex:
        print(ex)
        result = None
    finally:
        print("Excucations completed..")
    return result
        
myFunc(10,0)
```

```python
# Q2. Write a function that reads the contents of a file named `data.txt`. Use try, except, and finally blocks to 
# handle file not found errors and ensure the file is properly closed.

def file_reader(filename):
    try:
        with open(filename,'r') as file:
            content = file.read()
            print(content)
    except FileNotFoundError as e:
        print(f'error: {e}')
    except Exception as e:
        print(f'error: {e}')
    finally:
        print(f"Program excecutions completed.")
        
file_reader('data.txt')
```

```python
# Q3.Write a function that takes a list of integers and returns their sum. Use try, except, and finally blocks to 
# handle TypeError if a non-integer value is encountered and print an appropriate message.

def sum_of_list(lst):
    sum = 0
    try:
        for i in lst:
            sum += i
    except TypeError as e:
        print(f"error is : {e}") 
        sum = None
    except Exception as e:
        print(f"error: {e}")
        sum = None
    finally:
        print("Excecution is completed.")
    return sum
        
# sum_of_list([1,2,3,4,5])
sum_of_list([1,2,3,'4','a'])
```

```python
# Q4.Write a function that prompts the user to enter an integer. Use try, except, and finally blocks to handle 
# ValueError if the user enters a non-integer value and print an appropriate message.

def user_promts_handle():
    try:
        a = int(input("Enter the integer value: "))
        result = 10/a
    except ValueError as e:
        print(f"error: {e}")
        result = None
    except Exception as e:
        print(f"error: {e}")
        result = None
    finally:
        print("execution is completed.")
    return result

user_promts_handle()
```

```python
# Q5. Write a function that takes a dictionary and a key as input and returns the value associated with the key.
# Use try, except, and finally blocks to handle KeyError if the key is not found in the dictionary and print an 
# appropriate message.

def dict_func(dict1,key):
    try:
        value = dict1[key]
    except KeyError as e:
        print(f"error: {e}")
        value = None
    except Exception as e:
        print(f'error: {e}')
        value = None
    finally:
        print("Exceutiion is completed")
    return value

# dict_func({'a':1,'b':2,'c':3},'c')
print(dict_func({'a':1,'b':2,'c':3},'x'))
```

```python
# Q6. Write a function that performs nested exception handling. It should first attempt to convert a string to an
# integer, and then attempt to divide by that integer. Use nested try, except, and finally blocks to handle ValueError 
# and ZeroDivisionError and print appropriate messages.

def excep_func(k):
    try:
       try:
           num = int(k)
       except ValueError as e:
           print(f"Conversion error: {e}")
           num = None
       finally:
           print('Conversion program was successfully run.')
       if num is not None:
            try:
                result = 1000/num
            except ZeroDivisionError as e:
                print(f"Division error: {e}")
                result = None
            finally:
                print("Division program is successfully Completed.")   
            return result
    finally:
        print("Program overall completed")
        

# print(excep_func('7'))
print(excep_func('k'))
```

```python
# Q8.Write a function that attempts to open a URL and read its contents. Use try, except, and finally blocks to handle
# network-related errors and print an appropriate message.

import requests

def access_links(link):
    try:
        response = requests.get(link)
        response.raise_for_status()
        return response.text
    except requests.RequestException as e:
        print(f"Network error: {e}")
        return None
    finally:
        print("Execution Complete.")
        
# print(access_links('https://www.geeksforgeeks.org/python/python-programming-language-tutorial/'))
print(access_links('https://mahi.com'))
```

```python
# Q9. Write a function that attempts to parse a JSON string. Use try, except, and finally blocks to handle 
# JSONDecodeError if the string is not a valid JSON and print an appropriate message.

import json

def access_json(json_str):
    try:
        json_content = json.loads(json_str)
        return json_content
    except json.JSONDecodeError as e:
        print(f"Json file error: {e}")
        return None
    finally:
        print("Execution completed..")
        
# print(access_json('my invalid json'))
print(access_json('{"name": "John", "age": 30}'))
```

```python
# Q10.Define a custom exception named `NegativeNumberError`. Write a function that raises this exception if a negative
# number is encountered in a list. Use try, except, and finally blocks to handle the custom exception and print an 
# appropriate message.

class NegativeNumberError(Exception):
    pass

def postiveFunc(lst):
    try:
        for i in lst:
            if i < 0:
                raise NegativeNumberError(f"Negative number is found: {i}")
    except NegativeNumberError as e:
        print(f'error: {e}')
    finally:
        print("Execution Complete..")
        
print(postiveFunc([-1,2,3,-5,7]))
# print(postiveFunc([1,2,3,4,5]))
```

```python
# Q12. Define a class with a method that performs a division operation. Use try, except, and finally blocks within 
# the method to handle division by zero and print an appropriate message.

class Calculator:
    def divide(self,a,b):
        try:
            result = a/b
        except ZeroDivisionError as e:
            print(f"error: {e}")
            result = None
        finally:
            print("execution completed..")
        return result
    
obj = Calculator()
obj.divide(5,0)
```

```python
# Q13.Write a function that takes a list of strings and converts them to integers. Use try, except, and finally
# blocks to handle ValueError if a string cannot be converted and print an appropriate message.

def str_to_int(lst):
    l1 = []
    try:
        for item in lst:
            l1.append(int(item))
    except ValueError as e:
        print(f"error is: {e}")
        l1 = None
    finally:
        print("Program exceution is completed.")
    return l1

print(str_to_int(['1','2','3',4,'d']))
```

```python
# Q14.Write a function that uses a list comprehension to convert a list of strings to integers. Use try, except, and
# finally blocks within the list comprehension to handle ValueError and print an appropriate message.

def str_x_int(lst):
    try:
       l1 = [ int(i) for i in lst ]
    except ValueError as e:
        print(f"error is : {e}")
        l1 = None
    finally:
        print("Program execution is completed.")
    return l1

# str_x_int(['1','2',3,'4','5'])
str_x_int([3,'4','5','a','9'])
```

```python
# Q15.Write a function that attempts to write a list of strings to a file. Use try, except, and finally blocks to 
# handle IOError and ensure the file is properly closed.

def write_in_file(lst):
    try:
        with open('myone.txt','w') as file:
            for i in lst:
                file.write(i+ "\n")
                
    except IOError as e:
        print(f"error is: {e}")   
    finally:
        print("program executions is completed.")
        
write_in_file(["Connecto","shivam","is","Genius"])
```
