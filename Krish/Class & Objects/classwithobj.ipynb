```python
# Q3.Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit
# and withdraw money, and to check the balance. Create an object of the class and perform some operations.

class Bankaccounts:
    def __init__(self,account_number,balance):
        self.__account_number = account_number
        self.__balance = balance
        
    def diposite(self,amount):
        self.__balance += amount
        
    def withdraw(self,amount):
        if amount > self.__balance:
            print(f"Insuficient funds.")
        else:
            self.__balance -= amount
            
    def check_balance(self):
        return self.__balance
    

sbi = Bankaccounts(123,10000)
print(sbi.check_balance()) 
sbi.diposite(100)
print(sbi.check_balance()) 
sbi.withdraw(1000000)
print(sbi.check_balance()) 
sbi.withdraw(5000)
print(sbi.check_balance())
```

```python
# Q4. Create a base class named `Person` with attributes `name` and `age`. Create a derived class named `Employee` 
# that inherits from `Person` and adds an attribute `employee_id`. Create an object of the derived class and print 
# its attributes.

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
class Employee(Person):
    def __init__(self,name,age,emp_id):
        super().__init__(name,age)
        self.emp_id = emp_id
        
        
obj1 = Employee("sanskriti",21,123456)

print(obj1)
        
print(obj1.name)
print(obj1.age)
print(obj1.emp_id)
```

```python
# Q7. Create a class named `Counter` with a class variable `count`. Each time an object is created, increment the 
# count. Add a method to get the current count. Create multiple objects and print the count.

class Counter:
    count = 0
    def __init__(self):
        Counter.count += 1
        
    # @classmethod
    def get_count(self):
        return Counter.count 
    
obj1 = Counter()
obj2 = Counter()
obj3 = Counter()

print(obj3.get_count())
        
```

```python
# Q8.Create a class named `MathOperations` with a static method to calculate the square root of a number. Call the 
# static method without creating an object.
import math

class MathOperations:   
    @staticmethod
    def sqrt(num):
        return math.sqrt(num)
    
print(MathOperations.sqrt(15))
```

```python
# Q9.Create a class named `Rectangle` with private attributes `length` and `width`. Use properties to get and set 
# these attributes. Create an object of the class and test the properties.

class Rectangle:
    def __init__(self,length,width):
        self.__length = length
        self.__width = width
        
    @property
    def length(self):
        return self.__length
    
    @length.setter
    def length(self,length):
        self.__length = length
    
    @property
    def width(self):
        return self.__width
    
    @width.setter
    def width(self,width):
        self.__width = width
        
obj1 = Rectangle(5,6)
print(obj1.length ,obj1.width)
obj1.length = 8
obj1.width = 15
print(obj1.length,obj1.width)
```

```python
# Q10. Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` 
# and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

from abc import ABC,abstractmethod

class Shaped(ABC): 
    @abstractmethod
    def area(self):
        pass
        
        
class Circle(Shaped):
    def __init__(self,radius):
        self.radius = radius
        
    def area(self):
        return math.pi*self.radius**2
    

class Square(Shaped):
    def __init__(self,side):
        self.side = side
        
    def area(self):
        return self.side**2
   
obj_c1 = Circle(5)
print(obj_c1.area())
obj_s1 = Square(8)
print(obj_s1.area())
```

```python
# Q11.Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects.
# Create objects of the class and test the operator overloading.

class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y 
    def __add__(self,other):
        return Vector(self.x + other.x, self.y+other.y)
    
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
v1 = Vector(5,6)
v2 = Vector(9,8)
v3 = v1 + v2
print(v3)
```
