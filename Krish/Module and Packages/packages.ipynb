```python
# Q8.Use the `math` module to calculate the greatest common divisor (GCD) of two numbers and the factorial of a number.

import math

a  = math.gcd(48,18)
print(f'Gcd is : {a}')
print(f'factorial is : {math.factorial(a)}')
```

```python
# Q9.Use the `datetime` module to print the current date, calculate the date 100 days from today, 
# and determine the day of the week for a given date.

import datetime

today = datetime.date.today()
print(f"Today date is : {today}")

#Date 100 day from today
future_date = today + datetime.timedelta(days=100)
print(f"Date 100 days from today: {future_date}")

#Day of week of given date.
given_date = datetime.date(2023,2,21)
print(f'Day of week for 2023-02-21: {given_date.strftime('%A')}')
```

```python
# Q10.Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.

import random

random_numbers = [ random.randint(1,50) for _ in range(5)]
print(random_numbers)

# shuffle the list 
lst = [1,2,3,4,5]
random.shuffle(lst)
print(lst)
```
